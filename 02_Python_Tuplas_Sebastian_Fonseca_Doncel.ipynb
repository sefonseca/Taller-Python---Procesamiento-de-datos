# Examen sobre Tuplas

## Ejercicio 1

Considera la siguiente tupla de géneros musicales. Encuentra su longitud.


```python
# Tupla de géneros musicales

genres_tuple = ("pop", "rock", "soul", "hard rock", "soft rock",
                "R&B", "progressive rock", "disco")

print(len(genres_tuple))
```

### Explicación

La tupla contiene 8 elementos, por lo que len(genres_tuple) devuelve 8


## Ejercicio 2

Accede al elemento que se encuentra en el índice 3


```python
# Acceder al índice 3

print(genres_tuple[3])
```

### Explicación

El índice 3 corresponde al cuarto elemento porque Python comienza a contar desde 0


## Ejercicio 3

Utiliza slicing para obtener los elementos de los índices 3, 4 y 5


```python
# Obtener los índices 3, 4 y 5

print(genres_tuple[3:6])
```

### Explicación

El límite final del slicing no se incluye, por eso se utiliza 3:6 para obtener los índices 3, 4 y 5


## Ejercicio 4

Encuentra los primeros dos elementos de genres_tuple


```python
# Obtener los primeros dos elementos

print(genres_tuple[0:2])
```

### Explicación

El slicing 0:2 obtiene los elementos de los índices 0 y 1, que son "pop" y "rock"


## Ejercicio 5

Encuentra el primer índice donde aparece "disco"


```python
# Buscar el índice de "disco"

print(genres_tuple.index("disco"))
```

### Explicación

El método index() devuelve la posición de la primera aparición del valor indicado


## Ejercicio 6

Genera una lista ordenada a partir de la tupla C_tuple = (-5, 1, -3)


```python
# Convertir la tupla en una lista ordenada

C_tuple = (-5, 1, -3)
C_list = sorted(C_tuple)

print(C_list)
```

### Explicación

sorted() recibe la tupla y devuelve una lista ordenada de menor a mayor
