# Exámen sobre Listas

## Ejercicio 1

Crea una lista llamada a_list con los siguientes elementos: 1, hello, [1, 2, 3] y True.


```python
# Crear la lista solicitada

a_list = [1, "hello", [1, 2, 3], True]

print(a_list)
```

### Explicación

La lista contiene cuatro elementos de diferentes tipos,el tercer elemento es otra lista, por lo que se trata de una estructura anidada


## Ejercicio 2

Encuentra el valor almacenado en el índice 1 de a_list


```python
# Acceder al elemento del índice 1

print(a_list[1])
```

### Explicación

El índice 1 corresponde al segundo elemento de la lista, por lo tanto, el resultado es hello


## Ejercicio 3

Obtén los elementos almacenados en los índices 1, 2 y 3 de a_list


```python
# Obtener los elementos desde el índice 1 hasta el índice 3

print(a_list[1:4])
```

### Explicación

Se utiliza slicing para obtener los elementos de los índices 1, 2 y 3,el límite final del slicing no se incluye, por eso se utiliza 4


## Ejercicio 4

Concatena las siguientes listas:

A = [1, a]

B = [2, 1, d]


```python
# Crear las listas

A = [1, "a"]
B = [2, 1, "d"]

# Concatenar las listas

resultado = A + B

print(resultado)
```

### Explicación

El operador + permite concatenar dos listas, por lo que el resultado conserva el orden de los elementos de la primera lista y después agrega los elementos de la segunda lista
