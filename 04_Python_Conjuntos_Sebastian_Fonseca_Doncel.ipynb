# Cuestionario sobre conjuntos

## Ejercicio 1

Transforma la lista con los valores rap, house, electronic music y rap en un conjunto.


```python
# Convertir una lista en un conjunto

music_list = ["rap", "house", "electronic music", "rap"]
music_set = set(music_list)

print(music_set)
```

### Explicación

La función set() convierte la lista en un conjunto. Como los conjuntos no permiten elementos repetidos, el segundo valor rap se elimina automáticamente.


## Ejercicio 2

Considera la lista A = [1, 2, 2, 1] y el conjunto B = set([1, 2, 2, 1]). Comprueba que la suma de los elementos de A y B es diferente.


```python
# Crear la lista y el conjunto

A = [1, 2, 2, 1]
B = set([1, 2, 2, 1])

print("La suma de A es:", sum(A))
print("La suma de B es:", sum(B))
```

### Explicación

La lista conserva los cuatro elementos, incluyendo los valores repetidos, por lo que su suma es 6. El conjunto elimina los valores duplicados y conserva solamente 1 y 2, por lo que su suma es 3.


## Ejercicio 3

Crea un nuevo conjunto llamado album_set3 que sea la unión de album_set1 y album_set2


```python
# Crear los conjuntos

album_set1 = set(["Thriller", "AC/DC", "Back in Black"])
album_set2 = set(["AC/DC", "Back in Black", "The Dark Side of the Moon"])

# Crear la unión

album_set3 = album_set1.union(album_set2)

print(album_set3)
```

### Explicación

El método union() combina los elementos de los dos conjuntos. Los elementos que aparecen en ambos conjuntos solamente se almacenan una vez.


## Ejercicio 4

Averigua si album_set1 es un subconjunto de album_set3


```python
# Verificar si album_set1 es un subconjunto de album_set3

resultado = album_set1.issubset(album_set3)

print(resultado)
```

### Explicación

El método issubset() comprueba si todos los elementos de un conjunto también pertenecen al otro conjunto. En este caso, todos los elementos de album_set1 están dentro de album_set3, por lo que el resultado es True.
