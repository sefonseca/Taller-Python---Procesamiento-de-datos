# Cuestionario sobre Diccionarios

## Diccionario de ejemplo

Se utiliza el siguiente diccionario para las dos primeras preguntas:

soundtrack_dic = {"The Bodyguard": "1992", "Saturday Night Fever": "1977"}


## Ejercicio 1 a)

En el diccionario soundtrack_dic, encuentra cuáles son las llaves.


```python
# Crear el diccionario de ejemplo

soundtrack_dic = {
    "The Bodyguard": "1992",
    "Saturday Night Fever": "1977"
}

# Obtener las llaves

print(soundtrack_dic.keys())
```

### Explicación

El método keys() permite obtener todas las llaves del diccionario.

Las llaves son The Bodyguard y Saturday Night Fever


## Ejercicio 1 b)

En el diccionario soundtrack_dic, encuentra cuáles son los valores.


```python
# Obtener los valores

print(soundtrack_dic.values())
```

### Explicación

El método values() permite obtener todos los valores almacenados dentro del diccionario.

Los valores son 1992 y 1977


# Segundo grupo de ejercicios

Los discos Back in Black, The Bodyguard y Thriller poseen records de ventas en millones de 50, 50 y 65 respectivamente.


## Ejercicio 2 a)

Crea un diccionario llamado album_sales_dict donde las llaves sean los nombres de los discos y los valores sean las cantidades de ventas en millones.


```python
# Crear el diccionario de ventas

album_sales_dict = {
    "Back in Black": 50,
    "The Bodyguard": 50,
    "Thriller": 65
}

print(album_sales_dict)
```

### Explicación

Cada nombre de disco funciona como una llave y cada cantidad de ventas en millones funciona como su valor correspondiente.


## Ejercicio 2 b)

Usa el diccionario para encontrar las ventas totales de Thriller


```python
# Obtener las ventas de Thriller

print(album_sales_dict["Thriller"])
```

### Explicación

Se utiliza la llave Thriller para acceder al valor correspondiente. El resultado es 65 millones de ventas.


## Ejercicio 2 c)

Encuentra los nombres de los discos dentro del diccionario mediante el método keys


```python
# Obtener los nombres de los discos

print(album_sales_dict.keys())
```

### Explicación

El método `keys()` devuelve todas las llaves del diccionario. En este caso, muestra los nombres de los tres discos.


## Ejercicio 2 d)

Encuentra los records de ventas dentro del diccionario mediante el método values


```python
# Obtener los records de ventas

print(album_sales_dict.values())
```

### Explicación

El método values() devuelve todos los valores almacenados. En este caso, muestra 50, 50 y 65 millones.
