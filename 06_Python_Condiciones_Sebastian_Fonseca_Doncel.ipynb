# Cuestionario sobre Condiciones

## Ejercicio 1

Escribe una declaración if que determine si un disco musical tiene una calificación mayor a 8. Se utiliza la calificación del disco Back in Black, cuyo valor es 8.5. Si la condición es verdadera, imprime This album is Amazing.


```python
# Calificación del disco

rating = 8.5

if rating > 8:
    print("This album is Amazing!")
```

### Explicación

La calificación es 8.5 y es mayor que 8, por lo que la condición se cumple y se imprime el mensaje indicado.


## Ejercicio 2

Escribe una declaración if-else. Si la calificación es mayor que 8 imprime this album is amazing. Si es menor o igual a 8 imprime this album is ok.


```python
# Comprobar la calificación

rating = 8.5

if rating > 8:
    print("this album is amazing")
else:
    print("this album is ok")
```

### Explicación

Como la calificación es 8.5, se cumple la primera condición y se imprime this album is amazing.


## Ejercicio 3

Escribe una declaración if que determine si un disco fue lanzado antes de 1980 o en los años 1991 o 1993. Si la condición es verdadera, imprime el año de lanzamiento.


```python
# Año de lanzamiento

album_year = 1979

if album_year < 1980 or album_year == 1991 or album_year == 1993:
    print(album_year)
```

### Explicación

El año 1979 es anterior a 1980, por lo que la condición es verdadera y se imprime el año de lanzamiento.
