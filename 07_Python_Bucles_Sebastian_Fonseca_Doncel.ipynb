## Ejercicio 1

Escribe un bucle for que imprima todos los elementos entre -5 y 5 usando la función range.


```python
# Imprimir los valores entre -5 y 5

for i in range(-5, 6):
    print(i)
```

### Explicación

La función range genera los valores desde -5 hasta 5. El valor final de range no se incluye, por eso se utiliza 6.


## Ejercicio 2

Imprime todos los elementos de la lista de géneros musicales.


```python
# Lista de géneros musicales

Genres = ["rock", "R&B", "Soundtrack", "R&B", "soul", "pop"]

for Genre in Genres:
    print(Genre)
```

### Explicación

## Ejercicio 3

Escribe un bucle for que imprima la siguiente lista: squares = [red, yellow, green, purple, blue].


```python
# Lista de colores

squares = ["red", "yellow", "green", "purple", "blue"]

for square in squares:
    print(square)
```

### Explicación

## Ejercicio 4

Escribe un bucle while que muestre los valores de la calificación de las canciones almacenados en PlayListRatings. Si la calificación es menor a 6, sal del bucle. La lista es [10, 9.5, 10, 8, 7.5, 5, 10, 10].


```python
# Calificaciones de la lista de reproducción

PlayListRatings = [10, 9.5, 10, 8, 7.5, 5, 10, 10]

i = 0

while i < len(PlayListRatings):
    Rating = PlayListRatings[i]

    if Rating < 6:
        break

    print(Rating)
    i = i + 1
```

### Explicación

El bucle recorre las calificaciones desde el inicio. Cuando encuentra el valor 5, la condición Rating < 6 se cumple y el bucle termina.


## Ejercicio 5

Escribe un bucle while que copie las cadenas orange de la lista squares a la lista new_squares. El ciclo debe detenerse cuando encuentre un valor diferente de orange.


```python
# Copiar los valores orange hasta encontrar otro valor

squares = ["orange", "orange", "purple", "blue ", "orange"]
new_squares = []

i = 0

while i < len(squares) and squares[i] == "orange":
    new_squares.append(squares[i])
    i = i + 1

print(new_squares)
```

### Explicación

El ciclo agrega los valores orange a new_squares mientras encuentre ese mismo valor. Al encontrar purple, la condición deja de cumplirse y el ciclo termina.
