# Cuestionario sobre funciones

## Ejercicio 1

Crea una función que divida la primera entrada por la segunda entrada.


```python
# Crear una función para dividir dos valores

def div(a, b):
    return a / b

print(div(10, 2))
```

### Explicación

La función div recibe dos valores como parámetros y retorna el resultado de dividir el primero entre el segundo. Al utilizar 10 y 2, el resultado es 5.0.


## Ejercicio 2

Utiliza la función con para comprobar si se pueden agregar números enteros o cadenas.


```python
# Crear la función con

def con(a, b):
    return a + b

# Probar la función con números enteros

print(con(2, 2))

# Probar la función con cadenas

print(con("Hola ", "Python"))
```

### Explicación

La función con utiliza el operador +. Por eso puede sumar números enteros y también puede unir cadenas. Con los números 2 y 2 se obtiene 4, mientras que con las cadenas se obtiene Hola Python.


## Ejercicio 3

Utiliza la función con para comprobar si se puede concatenar una lista o una tupla.


```python
# Probar la función con listas

print(con(["a", 1], ["b", 1]))

# Probar la función con tuplas

print(con(("a", 1), ("b", 1)))
```

### Explicación

La función con también puede trabajar con listas y tuplas. En ambos casos, el operador + permite concatenar los elementos y devuelve una nueva estructura con los valores de las dos entradas.
