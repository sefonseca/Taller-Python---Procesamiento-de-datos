# Actividad a realizar

La clase matplotlib.patches.Ellipse permite dibujar elipses desde Matplotlib.

Se requiere crear una clase Elipse cuyo constructor tenga valores iniciales. La clase debe tener métodos para:

1. Cambiar el ancho.
2. Cambiar el alto.
3. Cambiar el color de relleno.
4. Cambiar el color de borde.
5. Dibujar.


```python
# Importar las librerías necesarias

import matplotlib.pyplot as plt
from matplotlib.patches import Ellipse
```

Crear la clase Elipse con valores iniciales para el ancho, el alto, el color de relleno y el color de borde.


```python
# Crear la clase Elipse

class Elipse:

    def __init__(self, width=4, height=2, facecolor="blue", edgecolor="black"):
        self.width = width
        self.height = height
        self.facecolor = facecolor
        self.edgecolor = edgecolor

    def cambiar_ancho(self, width):
        self.width = width

    def cambiar_alto(self, height):
        self.height = height

    def cambiar_color_relleno(self, facecolor):
        self.facecolor = facecolor

    def cambiar_color_borde(self, edgecolor):
        self.edgecolor = edgecolor

    def dibujar(self):
        figura, eje = plt.subplots()
        elipse = Ellipse(
            (0, 0),
            self.width,
            self.height,
            facecolor=self.facecolor,
            edgecolor=self.edgecolor
        )
        eje.add_patch(elipse)
        eje.set_xlim(-self.width, self.width)
        eje.set_ylim(-self.height, self.height)
        eje.set_aspect("equal")
        plt.show()
```

### Explicación

La clase Elipse contiene cuatro atributos: ancho, alto, color de relleno y color de borde. También contiene los cinco métodos solicitados para modificar estos atributos y dibujar la elipse.


Crear un objeto de la clase Elipse y comprobar sus valores iniciales.


```python
# Crear un objeto Elipse

mi_elipse = Elipse()

print("Ancho:", mi_elipse.width)
print("Alto:", mi_elipse.height)
print("Color de relleno:", mi_elipse.facecolor)
print("Color de borde:", mi_elipse.edgecolor)
```

### Explicación

Se crea un objeto utilizando los valores predeterminados del constructor. Estos valores permiten comprobar que los atributos fueron almacenados correctamente.



Utilizar los métodos para cambiar el ancho, el alto, el color de relleno y el color de borde.


```python
# Modificar los atributos de la elipse

mi_elipse.cambiar_ancho(8)
mi_elipse.cambiar_alto(4)
mi_elipse.cambiar_color_relleno("yellow")
mi_elipse.cambiar_color_borde("red")

print("Ancho:", mi_elipse.width)
print("Alto:", mi_elipse.height)
print("Color de relleno:", mi_elipse.facecolor)
print("Color de borde:", mi_elipse.edgecolor)
```

### Explicación

Los métodos de modificación cambian los atributos del objeto sin necesidad de crear una nueva elipse. Después de ejecutar los métodos, el ancho es 8, el alto es 4, el relleno es amarillo y el borde es rojo.



Utilizar el método para dibujar la elipse modificada.


```python
# Dibujar la elipse

mi_elipse.dibujar()
```

### Explicación

El método dibujar utiliza la clase Ellipse de Matplotlib para crear y mostrar la figura con los valores actuales del objeto.
